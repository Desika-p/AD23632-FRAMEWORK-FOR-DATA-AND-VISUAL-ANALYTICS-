Importing data from Database

```python
import sqlite3
import pandas as pd

# Connect to (or create) the database
conn = sqlite3.connect('my_database.db')
cursor = conn.cursor()

# Create the 'students' table
cursor.execute('''
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    major TEXT,
    gpa REAL,
    enrollment_date TEXT
)
''')

# Insert example records
cursor.executemany('''
INSERT INTO students (id, name, major, gpa, enrollment_date) VALUES (?, ?, ?, ?, ?)
''', [
    (1, 'Nina',      'Computer Science',        3.8, '2021-08-15'),
    (2, "Lerin",   'Mechanical Engineering',  3.4, '2020-08-20'),
    (3, 'Sofia',   'AIML',                 3.9, '2022-01-10'),
    (4, 'Krithick',        'AIDS',               3.2, '2019-08-25'),
    (5, 'Bhagat',  'Computer Science',         3.6, '2021-01-12'),
])

# Commit and close
conn.commit()
print("Database and 'students' table created with sample data.")

# Query into a DataFrame
sql_df = pd.read_sql_query("SELECT * FROM students", conn)
print(sql_df)
```

**Webscraping data**

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36"
}

response = requests.get(url, headers=headers, timeout=10)
response.raise_for_status()   # will raise a clear error instead of silently returning a 403 page

soup = BeautifulSoup(response.content, "html.parser")
html_table = soup.find("table", {"class": "wikitable"})

if html_table is None:
    raise ValueError("No wikitable found — page structure may have changed")

web_df = pd.read_html(str(html_table))[0]
print(web_df.head())
```
